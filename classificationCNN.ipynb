```python
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Conv1D
from tensorflow.keras.layers import GlobalMaxPooling1D, Dense

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
texts = [
    "the movie was excellent",
    "i really enjoyed the film",
    "the acting was amazing",
    "this movie was fantastic",
    "the movie was terrible",
    "i did not enjoy the film",
    "the acting was awful",
    "this movie was boring"
]

labels = [1, 1, 1, 1, 0, 0, 0, 0]
#1 = Positive
#0 = Negative
```

```python
tokenizer = Tokenizer(num_words=1000)

tokenizer.fit_on_texts(texts)

sequences = tokenizer.texts_to_sequences(texts)

print("Word Index:")
print(tokenizer.word_index)

print("\nSequences:")
print(sequences)
```

```python
#Padding
max_length = 10

X = pad_sequences(
    sequences,
    maxlen=max_length,
    padding='post'
)

y = np.array(labels)

print(X)
```

```python
#Build the CNN Model
model = Sequential([
    Embedding(1000, 50, input_length=max_length),

    Conv1D(
        filters=32,
        kernel_size=3,
        activation="relu"
    ),

    GlobalMaxPooling1D(),

    Dense(
        1,
        activation="sigmoid"
    )
])

model.summary()
```

```python
#Compile the Model
model.compile(
optimizer='adam',
loss='binary_crossentropy',
metrics=['accuracy']
)
```

```python
#Train the Model
model.fit(
X,
y,
epochs=100,
verbose=1
)
```

```python
#Test the Model
test_texts = [
"the movie was fantastic",
"the movie was horrible"
]
test_sequences = tokenizer.texts_to_sequences(
test_texts)

test_data = pad_sequences(
test_sequences,
maxlen=max_length,
padding='post')

predictions = model.predict(test_data)

for review, prediction in zip(test_texts, predictions):
    sentiment = "Positive" if prediction > 0.5 else "Negative"
    print(f"Review: {review}")
    print(f"Score: {prediction[0]:.4f}")
    print(f"Prediction: {sentiment}\n")
```
